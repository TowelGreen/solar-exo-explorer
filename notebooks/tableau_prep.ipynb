```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("merged_planetary_data.csv")
```

```python
data.columns
```

```python
# Create Sun reference row
sun_data = {
    'pl_name': 'SunRef',
    'pl_orbper': np.nan,
    'pl_orbsmax': np.nan,
    'pl_orbeccen': np.nan,
    'pl_rade': np.nan,
    'pl_bmasse': np.nan,
    'st_teff': 5778,
    'st_rad': 1.0,
    'st_mass': 1.0,
    'st_met': 0.0,
    'origin': 'Sun'
}

# Append to combined_df
data = pd.concat([data, pd.DataFrame([sun_data])], ignore_index=True)
```

```python
def habitability_score(row):
    score = 0
    if 0.5 <= row['pl_bmasse'] <= 2.0:
        score += 1
    if 180 <= row['pl_eqt'] <= 310:
        score += 1
    if 0.35 <= row['pl_insol'] <= 1.75:
        score += 1
    return score / 3  # Normalize to 0–1
```

```python
def is_habitable(row):
    return "Yes" if row['habitability_score'] == 1.0 else "No"
```

```python
# Apply to your merged dataframe
data['habitability_score'] = data.apply(habitability_score, axis=1)
data['habitable'] = data.apply(is_habitable, axis=1)
```
